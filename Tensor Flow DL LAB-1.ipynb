# Experiment 3: Basic Tensor Operations

```python
tensor1 = tf.constant([10, 20, 30])
tensor2 = tf.constant([5, 10, 15])

addition = tf.add(tensor1, tensor2)
subtraction = tf.subtract(tensor1, tensor2)
multiplication = tf.multiply(tensor1, tensor2)
division = tf.divide(tensor1, tensor2)

print("\nAddition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

# Experiment 4: Data Types and Shape of Tensors

```python
tensor = tf.constant([[1, 2, 3],
                      [4, 5, 6]])

print("\nTensor:")
print(tensor)

print("Data Type:", tensor.dtype)
print("Shape:", tensor.shape)
print("Number of Dimensions:", tensor.ndim)
```

# Experiment 5: Converting NumPy Arrays to Tensors

```python
import numpy as np
```

```python
numpy_array = np.array([1, 2, 3, 4, 5])
tf_tensor = tf.convert_to_tensor(numpy_array)

print("\nNumPy Array:")
print(numpy_array)

print("\nConverted TensorFlow Tensor:")
print(tf_tensor)
```
